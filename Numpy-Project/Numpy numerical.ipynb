```python

import numpy as np
from PIL import Image
import matplotlib.pyplot as plt

img = Image.open("radhakrishna.jpg").convert("RGB")


img_arr = np.array(img)


img_arr[250:, 370:, 1] = 0
img_arr[0:250, 0:370, 2] = 0
img_arr[0:250, 370:, 0] = 0
img_arr[250:, 0:370, 0] = 250


plt.figure(figsize=(10, 8))
plt.imshow(img_arr)

plt.axis("on")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")

plt.show()


```
