```python
import matplotlib.pyplot as plt
```

```python
x=[1,2,3,4,5,6]
y=[100,200,250,300,350,200]
```

# Line chart

```python
plt.plot(x,y)
plt.show()
```

```python
plt.plot(x,y)
plt.title("Sales Report")
plt.xlabel("Days")
plt.ylabel("Unit Sell")
plt.show()
```

```python
plt.plot(x,y, color="green")
plt.title("Sales Report")
plt.xlabel("Days")
plt.ylabel("Unit Sell")
plt.show()
```

# Bar graph

```python
x=[1,2,3,4,5,]
y=["Monday","Tuesday","Wednesday","Thursday","Friday"]
```

```python
plt.bar(x,y)
plt.title("day wise sales")
plt.show()
```

# Horizontal bar graph

```python
plt.barh(x,y)
plt.title("day wise sales")
plt.show()
```

# pie chart

```python
x=['a','b','c','d','e','f']
y=[1,2,3,4,5,2]
plt.pie(y, labels=x,autopct ="%1.1f%%")
plt.show()
```

# Histogram 

```python
x=[10,12,15,18,20,21,22,24,25]
plt.hist(x, bins=2)
plt.show()
```

# scatter plot

```python
x=[1,2,3,4,]
y=[5,8,6,10]
plt.scatter(x,y)
plt.show()
```

# violin plot

```python
x=[[10,20,30,40,50]]
plt.violinplot(x)
plt.show()
```

# area chart

```python
x=[1,2,3,4,1,9,8]
y=[100,200,188,353,2556,1234,242]
plt.fill_between(x,y)
plt.show()
```

# Stack plot

```python
x=[1,2,3]
y1=[2,3,4]
y2=[1,2,3]
y3=[5,6,7]
plt.stackplot(x,y1,y2,y3)
plt.show()
```

# stem plot

```python
x=[1,2,3]
y=[2,3,4]
plt.stem(x,y)
plt.show()
```

# step chart

```python
plt.step([1,2,3,4],[5,8,6,9])
plt.show()
```

# heatmap

```python
x=([1,2,3,4],[5,8,6,9])
plt.imshow(x)
plt.colorbar()
plt.show()
```
